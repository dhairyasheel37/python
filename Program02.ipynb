# 6. Add 100 to the copied list and explain why the original list does not change.
# 7. Calculate the total of the original list.
# 8. Create a tuple containing the student's name and total marks.
 #9. Create a set from the marks list to show only unique marks.
# 10. Create a dictionary containing:
#        "name"   → student's name
#        "marks"  → original marks list
#        "total"  → total marks
#        "unique" → set of unique marks
# 11. Print the final result using an f-string.

```python
name = input("Enter the student name :")
L1 = []
for i in range(3):
    I = int(input(f"Enter the List "))
    L1.append(I) 

L1.append(90)
L1copy = L1.copy()

L1copy.append(100)
# Orignal L List not change here Because we use copy. Reference Value changed Here
sum = 0
for i in range(4):
    sum += L1[i]

print(f"Original of sum : {sum}")
tuple1 = (name,sum)
set1 = set(L1)

dict1 = {"name":name,"marks":L1,"total":sum,"Unique":set1}
print(dict1)
print(set1)
print(tuple1)
print(L1)
print(L1copy)
```
